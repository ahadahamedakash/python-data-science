```python
import json
```

```python
# load the data
def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)

    return data
```

```python
data = load_data("data.json")
```

```python
# get meaningful insights from data
def get_insights(data):
    # avg rating
    total_rating = 0
    for d in data:
        total_rating += float(d["rating"])

    print(f"Avg Rating: {total_rating / len(d)}")

    # percentage of user with poor rating
    poor_ratings = 0
    for d in data:
        if(float(d["rating"]) < 3):
            poor_ratings += 1

    print(f"Percentage of user with poor rating: {poor_ratings / len(data) * 100}%")
```

```python
get_insights(data)
```
