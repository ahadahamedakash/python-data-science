```python
import numpy as np
```

```python
arr = np.array([[1, 2], [3, 4]])

mean_val = np.mean(arr)  # (1 + 2 + 3 + 4) / 4 = 2.5

std_dev = np.std(arr)  # sqrt(((1-2.5)^2 + (2-2.5)^2 + (3-2.5)^2 + (4-2.5)^2) / 4)

print(mean_val)
print(std_dev)

normalized_arr = (arr - mean_val) / std_dev

print(normalized_arr)
```
