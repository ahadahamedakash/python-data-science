```python
import numpy as np
```

```python
arr = np.array([1, 2, 3, 4, 5])

# Aggregation
print(np.sum(arr))       # Calculates the sum of all elements
print(np.prod(arr))      # Calculates the product of all elements
print(np.min(arr))       # Returns the minimum value
print(np.max(arr))       # Returns the maximum value
print(np.mean(arr))      # Calculates the arithmetic mean (average)
print(np.std(arr))       # Calculates the standard deviation
print(np.var(arr))       # Calculates the variance
print(np.median(arr))    # Returns the median value
print(np.argmin(arr))    # Returns the index of the minimum value
print(np.argmax(arr))    # Returns the index of the maximum value

# Power
print(np.square(arr))    # Squares each element (x²)
print(np.sqrt(arr))      # Calculates the square root of each element
print(np.power(arr, 2))  # Raises each element to the power of 2

# Log & Exponential
print(np.log(arr))       # Calculates the natural logarithm (base e)
print(np.log10(arr))     # Calculates the base-10 logarithm
print(np.log2(arr))      # Calculates the base-2 logarithm
print(np.exp(arr))       # Calculates e raised to the power of each element (eˣ)

# Rounding
print(np.round(arr))     # Rounds each element to the nearest integer
print(np.floor(arr))     # Rounds each element down to the nearest integer
print(np.ceil(arr))      # Rounds each element up to the nearest integer
print(np.trunc(arr))     # Removes the decimal part (toward zero)

# Extras
print(np.unique(arr))    # Returns sorted unique elements (removes duplicates)
print(np.sort(arr))      # Returns a sorted copy of the array
print(np.abs(arr))       # Returns the absolute value of each element
```
