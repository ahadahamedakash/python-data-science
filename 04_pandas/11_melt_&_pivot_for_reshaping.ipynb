```python
import pandas as pd
```

```python
# melt and pivot

df = pd.DataFrame({
    "country" : ["USA", "UK", "CH", "BA"],
    "year" : [2020, 2021, 2020, 2021],
    "sales" : [100, 250, 180, 220],
    "profit" : [20, 10, 10, 30],
})

df
```

```python
# melt(id_vars,  value_vars,    var_name,         value_name)
#      (keep)     (unpivot)  (new col name)     (new col value)  

melted_df = df.melt(
    id_vars = ["country", "year"],
    value_vars = ["sales", "profit"],
    var_name="metrics",
    value_name = "value"
)

melted_df
```

```python
# pivot

# pivot(index, columns, values)
#       (keep)

original = melted_df.pivot(
    index=["country", "year"],
    columns="metrics",
    values="value"
)

original
```
