```python
import pandas as pd
```

```python
info = {
    "name" : ["Rana", "Rahim", "Rafi"],
    "age" : [35, 50, 25]
}

df = pd.DataFrame(info)

print(df)
```

```python
# series (1D labeled array)
s = pd.Series([1, 2, 3, 4, 5])

print(s)
```

```python
s = pd.Series([1, 2, 3, 4], index=["rahim", "karim", "jabbar", "shaikot"])

print(s)

print(s["jabbar"])

print(s.iloc[0])
```

```python
# Series (properties)

# 1. Homogenous
# 2. Vectorized
# 3. Handle Missing Values with NaN
# 4. Mutable values, immutable size
```

```python
s = pd.Series([1, 2, 3, 4, "test"], index=["rahim", "karim", "jabbar", "shaikot", 5]) # series are homogenous

print(s)

print(s.index)
```

```python
s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([6, 7, 8, 9, 0])

print(s1 + s2)

s1[0]  = 100

print(s1)

dr = s1.drop(0)

print(dr)
```
