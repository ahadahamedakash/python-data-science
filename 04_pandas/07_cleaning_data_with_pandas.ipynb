```python
import pandas as pd
```

```python
# import data

df = pd.read_csv("raw_data.csv")

# Check missing values
df.isnull()  
# Checks every cell for missing values.
# Returns a Boolean DataFrame (True = missing, False = not missing).

df.isna()  
# Alias of isnull(). Performs exactly the same operation.

df.isnull().sum()  
# Counts the number of missing values in each column.
# Returns a Pandas Series containing column names and their missing value counts.

df.dropna()       
# Removes rows containing at least one missing value.
# Returns a new DataFrame.
# Does not modify the original DataFrame by default.

df.dropna(axis=1) 
# Removes columns containing at least one missing value.
# axis=0 means rows, axis=1 means columns.
# Returns a new DataFrame.
# Does not modify the original DataFrame by default.

cleaned_data = df.copy()  
# Creates an independent copy of the original DataFrame.
# Changes to cleaned_data will not affect df.


age_mean = cleaned_data["age"].mean()  
# Calculates the average age from the copied DataFrame.


cleaned_data["age"] = cleaned_data["age"].fillna(age_mean)
# Replaces missing values in the age column with the calculated mean.
# Assigns the updated Series back to the age column.
# Modifies cleaned_data, but leaves the original df unchanged.

df.ffill()
# Forward Fill:
# Replaces missing values with the previous valid value.
# Returns a new DataFrame.
# Does not modify the original DataFrame by default.


df.bfill()
# Backward Fill:
# Replaces missing values with the next valid value.
# Returns a new DataFrame.
# Does not modify the original DataFrame by default.

# df.fillna({
#     "age": 0
# })
# This code will not execute because it is commented out.

df
```

```python
# Handle duplicate

df.duplicated()
# Checks whether each row is a duplicate of a previous row.
# Returns a Boolean Series (True = duplicate, False = unique).
# The first occurrence of a row is considered unique (False).
# Does not modify the original DataFrame.


df.duplicated().sum()
# # Counts the total number of duplicate rows.
# # Returns an integer value.
# # Does not modify the original DataFrame.

df.drop_duplicates()
# Removes duplicate rows from the DataFrame.
# Keeps the first occurrence by default.
# Returns a new DataFrame.
# Does not modify the original DataFrame by default.

df.duplicated(keep="first")
# Marks the first occurrence as False.
# Marks subsequent duplicates as True.
# Default behavior.

df.duplicated(keep="last")
# Marks the last occurrence as False.
# Marks previous duplicates as True.

df.duplicated(keep=False)
# Marks ALL occurrences of duplicate rows as True.
# No occurrence is treated as unique.

df.drop_duplicates(keep="first")
# Keeps the first occurrence and removes subsequent duplicates.

df.drop_duplicates(keep="last")
# Keeps the last occurrence and removes previous duplicates.

df.drop_duplicates(keep=False)
# Removes ALL rows that have duplicates.
# Only rows that appear exactly once are retained.

df.duplicated(subset=["name"])
# Checks duplicates using only the name column.
# Returns a Boolean Series.

df.drop_duplicates(subset=["name"])
# Removes rows with duplicate names.
# Keeps the first occurrence by default.
# Returns a new DataFrame.

df.drop_duplicates(inplace=True)
# Removes duplicate rows directly from the original DataFrame.
# Modifies df itself.
# Returns None instead of a new DataFrame.

df
```
