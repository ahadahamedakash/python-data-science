```python
import pandas as pd
```

```python
# import data

df = pd.read_csv("raw_data.csv")

df
```

```python
# apply
df["age"].apply(lambda x: x + 1)  # Applies a function to each value and returns a Series.

# apply with a custom function

def calculate_tax(income):
    return income * 0.10

df["income"].apply(calculate_tax)  # Applies the custom function to each income value and returns a Series containing calculated tax.

# map
df["gender"].map({"Male": 1, "Female": 0})  # Maps values using a dictionary or function and returns a Series.

# assign
df.assign(age=df["age"] + 1)  # Creates a new DataFrame with a new or updated column and returns the DataFrame.

# replace(old, new)
df["gender"].replace("Male", "M")  # Replaces matching old values with new values and returns a Series.
```

```python
def calculate_tax(income):
    return income * 0.15

df["tax"] = df["income"].apply(calculate_tax)

df[["name", "tax"]]
```

```python
# rename
df.rename(columns={"name": "full_name"})  # Renames the specified column(s) and returns a new DataFrame.
df.rename(index={1 : "First"}) # rename index

# sort_values
df.sort_values(by="income")  # Sorts rows based on column values in ascending order and returns a new DataFrame.
df.sort_values(["income", "age"]) # sort by two values

# sort_values (descending)
df.sort_values(by="income", ascending=False)  # Sorts rows in descending order based on income and returns a new DataFrame.

# sort_index
df.sort_index()  # Sorts rows based on their index labels in ascending order and returns a new DataFrame.

# reset_index
df.reset_index()  # Resets the DataFrame index to the default integer index and returns a new DataFrame.
df.reset_index(drop=True)

# rank
df["income"].rank()  # Assigns ranks to values in ascending order and returns a Series.
df["income"].rank(ascending=False, method="dense") # instead of dense we can use min and max

# reorder columns
df[["name", "age", "gender", "country", "income", "tax"]]  # Reorders columns in the specified order and returns a new DataFrame.
```
