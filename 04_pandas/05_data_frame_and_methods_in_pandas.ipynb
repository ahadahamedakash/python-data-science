```python
# Creating a sample DataFrame
data = {
    "name": ["Rahim", "Karim", "Jabbar", "Shaikot", "Nabil", "Hasan", "Rakib", "Sakib"],
    "age": [25, 30, 28, 22, 35, 30, 25, 28],
    "city": ["Dhaka", "Chittagong", "Dhaka", "Narayanganj", "Sylhet", "Dhaka", "Sylhet", "Narayanganj"],
    "salary": [25000, 35000, 30000, 20000, 45000, 35000, 28000, 32000],
    "department": ["IT", "HR", "IT", "Sales", "Finance", "HR", "IT", "Sales"]
}

df = pd.DataFrame(data)


# 1. df.head()
print(df.head())  # Returns the first 5 rows by default; output shows the beginning of the DataFrame.


# 2. df.tail()
print(df.tail())  # Returns the last 5 rows by default; output shows the ending of the DataFrame.


# 3. df.sample()
print(df.sample())  # Returns 1 random row by default; output shows a randomly selected record.


# 4. df.info()
df.info()  # Displays column names, data types, non-null counts and memory usage; output is a summary, not a DataFrame.


# 5. df.shape
print(df.shape)  # Returns a tuple (rows, columns); output is (8, 5).


# 6. df.describe()
print(df.describe())  # Returns descriptive statistics of numeric columns, including count, mean, std, min, max and quartiles.


# 7. df.columns
print(df.columns)  # Returns all column names as a pandas Index object.


# 8. df.nunique()
print(df.nunique())  # Returns the number of unique values in each column; output is a count for every column.
```
