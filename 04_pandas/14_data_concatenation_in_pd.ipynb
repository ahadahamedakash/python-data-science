```python
import pandas as pd
```

```python
df1 = pd.DataFrame({
    "id": [1, 2, 3],
    "name": ["adam", "bob", "jhon"]
})

df2 = pd.DataFrame({
    "id": [11, 12, 13],
    "name": ["charlie", "gilian", "fester"]
})

pd.concat([df1, df2]) # row
pd.concat([df1, df2], ignore_index=True) # row

pd.concat([df1, df2], axis=1) # column
```
